# 🔢 One Hot Encoding — Complete Guide

**One Hot Encoding** converts categorical variables into a binary (0/1) matrix format.  
Each unique category becomes a new column, with `1` indicating the presence and `0` indicating absence.

---
### Why One Hot Encoding?
- ML models work with **numbers**, not text
- Unlike Label Encoding, OHE avoids imposing **ordinal relationships** between categories
- Best for **nominal** (unordered) categories like Color, City, Gender

---

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

print("✅ Libraries imported successfully!")
```

## 🗂️ Step 2 — Create Dummy Dataset

We'll use a **Customer Purchase Dataset** with the following categorical columns:
- `Gender` → Male / Female
- `City` → Mumbai / Delhi / Bangalore / Chennai
- `Product_Category` → Electronics / Clothing / Food / Sports
- `Membership` → Gold / Silver / Bronze

```python
# ── Dummy Customer Dataset ──────────────────────────────────────────
data = {
    'CustomerID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Gender':     ['Male', 'Female', 'Female', 'Male', 'Male',
                   'Female', 'Male', 'Female', 'Male', 'Female'],
    'City':       ['Mumbai', 'Delhi', 'Bangalore', 'Chennai', 'Mumbai',
                   'Delhi', 'Bangalore', 'Chennai', 'Mumbai', 'Delhi'],
    'Product_Category': ['Electronics', 'Clothing', 'Food', 'Sports', 'Electronics',
                         'Food', 'Clothing', 'Sports', 'Food', 'Electronics'],
    'Membership': ['Gold', 'Silver', 'Bronze', 'Gold', 'Bronze',
                   'Silver', 'Gold', 'Bronze', 'Silver', 'Gold'],
    'Purchase_Amount': [15000, 3200, 850, 5400, 22000,
                        1100, 7800, 4300, 960, 18500]
}

df = pd.DataFrame(data)
print("📋 Original Dataset:")
print(df.to_string(index=False))
print(f"\nShape: {df.shape}")
print(f"\nData Types:\n{df.dtypes}")
```

## 🔍 Step 3 — Explore Unique Categories

```python
categorical_cols = ['Gender', 'City', 'Product_Category', 'Membership']

print("📊 Unique values in each categorical column:")
print("-" * 45)
for col in categorical_cols:
    uniques = df[col].unique().tolist()
    print(f"  {col:20s} → {uniques}")
```

---
## 🛠️ Method 1 — One Hot Encoding using `sklearn`

`OneHotEncoder` from `sklearn.preprocessing` is the most widely used approach in ML pipelines.

```python
# ── sklearn OneHotEncoder ───────────────────────────────────────────
encoder = OneHotEncoder(sparse_output=False, dtype=int)

# Fit and transform the categorical columns
encoded_array = encoder.fit_transform(df[categorical_cols])

# Get the generated feature names
feature_names = encoder.get_feature_names_out(categorical_cols)

# Build a DataFrame from the encoded array
df_encoded_sklearn = pd.DataFrame(encoded_array, columns=feature_names)

print("✅ sklearn OneHotEncoder Output:")
print(f"   New columns created: {len(feature_names)}")
print(f"   Feature names: {list(feature_names)}\n")
print(df_encoded_sklearn.to_string(index=False))
```

```python
# ── Combine with original numerical columns ─────────────────────────
df_final_sklearn = pd.concat(
    [df[['CustomerID', 'Purchase_Amount']].reset_index(drop=True),
     df_encoded_sklearn],
    axis=1
)

print("🔗 Final Encoded Dataset (sklearn):")
print(f"   Shape: {df_final_sklearn.shape}")
print(df_final_sklearn.to_string(index=False))
```

---
## 🐼 Method 2 — One Hot Encoding using `pandas.get_dummies()`

The simplest and most readable approach — great for quick EDA and small datasets.

```python
# ── pandas get_dummies ──────────────────────────────────────────────
df_pandas_ohe = pd.get_dummies(
    df,
    columns=categorical_cols,
    dtype=int
)

print("✅ pandas get_dummies Output:")
print(f"   Shape: {df_pandas_ohe.shape}")
print(f"   Columns: {list(df_pandas_ohe.columns)}\n")
print(df_pandas_ohe.to_string(index=False))
```

---
## ⚡ Method 3 — One Hot Encoding using `numpy` (Manual)

Understanding the inner workings — manually creating the binary matrix using NumPy.

```python
# ── Manual OHE with NumPy ───────────────────────────────────────────
def manual_one_hot(series):
    """Manually apply One Hot Encoding to a pandas Series using NumPy."""
    categories = sorted(series.unique())
    ohe_matrix = np.zeros((len(series), len(categories)), dtype=int)
    cat_to_idx = {cat: idx for idx, cat in enumerate(categories)}
    for row_idx, value in enumerate(series):
        ohe_matrix[row_idx, cat_to_idx[value]] = 1
    col_names = [f"{series.name}_{cat}" for cat in categories]
    return pd.DataFrame(ohe_matrix, columns=col_names)

# Apply to each categorical column and concatenate
encoded_parts = [df[['CustomerID', 'Purchase_Amount']].reset_index(drop=True)]
for col in categorical_cols:
    encoded_parts.append(manual_one_hot(df[col]))

df_numpy_ohe = pd.concat(encoded_parts, axis=1)

print("✅ Manual NumPy OHE Output:")
print(f"   Shape: {df_numpy_ohe.shape}")
print(df_numpy_ohe.to_string(index=False))
```

---
## 🧠 Key Takeaways

| Concept | Detail |
|---|---|
| **When to use OHE** | For **nominal** categories with no natural order |
| **Dummy Variable Trap** | Use `drop_first=True` in `get_dummies` to avoid multicollinearity |
| **High Cardinality** | OHE can create too many columns — consider Target Encoding instead |
| **sklearn advantage** | Can `transform()` new unseen data using the fitted encoder |

```python
# ── BONUS: Drop First to Avoid Dummy Variable Trap ──────────────────
print("⚠️  Avoiding the Dummy Variable Trap (drop_first=True):\n")

df_no_trap = pd.get_dummies(
    df,
    columns=categorical_cols,
    drop_first=True,
    dtype=int
)

print(f"   Shape WITH  all dummies : {df_pandas_ohe.shape}")
print(f"   Shape WITHOUT 1st dummy : {df_no_trap.shape}")
print(f"\n   Columns after drop_first:")
print(f"   {list(df_no_trap.columns)}")
```
